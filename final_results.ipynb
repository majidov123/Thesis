```python
df_runs = pd.DataFrame(runs)

average_results = (
    df_runs
    .groupby(["Model", "Strategy"], as_index=False)
    .mean(numeric_only=True)
)

average_results = average_results.drop(columns=["Run"])

average_results["Model_Strategy"] = (
    average_results["Model"] + "_" + average_results["Strategy"]
)

average_results
```

```python
final_matrix = average_results.set_index("Model_Strategy").drop(columns=["Model", "Strategy"])

final_matrix
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(final_matrix, annot=True, cmap="coolwarm", fmt=".4f")
plt.title("Average Evaluation Metrics for Causal Discovery Models")
plt.xlabel("Evaluation Metrics")
plt.ylabel("Model + Prompt Strategy")
plt.tight_layout()
plt.show()
```
